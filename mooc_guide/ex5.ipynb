```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.io.arff import loadarff 
```

### categorical data
categorical variables have discrete values

```python
data, metadata = loadarff('../datasets/adult_census.arff')
adult_census = pd.DataFrame(data)
for col in adult_census.select_dtypes(include='object'):
    adult_census[col] = adult_census[col].str.decode('utf-8')
adult_census = adult_census.drop(columns='education-num')

data = adult_census.drop(columns='class')
target = adult_census['class']
data.head()
```

```python
data[data.dtypes[data.dtypes=='str'].index].head()
```

```python
from sklearn.compose import make_column_selector as selector

categorical_cols_selector = selector(dtype_include=str)
categorical_cols = categorical_cols_selector(data)
categorical_cols
```

```python
data_categorical = data[categorical_cols]
data_categorical
```

no encoding

### encoding categories

#### OneHotEncoder -> linear models 
#### OrdinalEncoder -> tree-based models

* ordinal categories
--> encode each category with a different number

```python
from sklearn.preprocessing import OrdinalEncoder

education_column = data_categorical[['education']]
encoder = OrdinalEncoder().set_output(transform='pandas')
education_encoded = encoder.fit_transform(education_column)
education_encoded
```

```python
data_encoded = encoder.fit_transform(data_categorical)
data_encoded.head()
```

* encoding nominal categories (without assuming any order)

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False).set_output(transform="pandas")
education_encoded = encoder.fit_transform(education_column)
education_encoded.iloc[0]
```

```python
data_encoded = encoder.fit_transform(data_categorical)
data_encoded.head()
```

### evaluating

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_validate
from sklearn.pipeline import Pipeline

# using OneHotEncoder
pipe = Pipeline([('encoder', OneHotEncoder(handle_unknown='ignore')), ('model', LogisticRegression(max_iter=500))])
cv_score = cross_validate(pipe, data_categorical, target)['test_score']
f'accuracy is {cv_score.mean():.4f} ± {cv_score.std():.4f}'
```

```python
pipe = Pipeline([('encoder', OneHotEncoder(min_frequency=25, handle_unknown='infrequent_if_exist')), ('model', LogisticRegression(max_iter=500))])
cv_score = cross_validate(pipe, data_categorical, target)['test_score']
f'accuracy is {cv_score.mean():.4f} ± {cv_score.std():.4f}'
```

### ordinal encoder vs one hot encoder

```python
pipe.named_steps
```

```python
from sklearn.base import clone

def comparison_report(model):
    pipe_ordinal = Pipeline([('encoder', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)), ('model', clone(model))])
    pipe_one_hot = Pipeline([('encoder', OneHotEncoder(handle_unknown='ignore')), ('model', clone(model))])
    cv_ordinal = cross_validate(pipe_ordinal, data_categorical, target, error_score="raise")
    cv_one_hot = cross_validate(pipe_one_hot, data_categorical, target, error_score="raise")
    print(f"{pipe_ordinal.named_steps['encoder']} -> {cv_ordinal['test_score'].mean():.4f} ± {cv_ordinal['test_score'].std():.4f}")
    print(f"{pipe_one_hot.named_steps['encoder']} -> {cv_one_hot['test_score'].mean():.4f} ± {cv_one_hot['test_score'].std():.4f}")
```

* logistic regression

```python
comparison_report(LogisticRegression(max_iter=500))
```

* random forest

```python
from sklearn.ensemble import RandomForestClassifier

comparison_report(RandomForestClassifier(n_estimators=100))
```

* decision tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier

model = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=1),
    n_estimators=100,
    random_state=42
)

comparison_report(model)
```
